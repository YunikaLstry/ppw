# Contoh Skipgram

```python
from gensim.models import Word2Vec
import pandas as pd
```

```python
# 1. Membuat 3 kalimat
sentences = [
    "Kucing tidur di atas sofa",
    "Anak membaca buku di perpustakaan",
    "Mobil merah melaju dengan cepat"
]

# 2. Tokenisasi dan lowercase
tokenized_sentences = [
    sentence.lower().split()
    for sentence in sentences
]

print("Hasil tokenisasi:")
print("-" * 40)
for sentence in tokenized_sentences:
    print(sentence)

# 3. Membuat model Word2Vec
model = Word2Vec(
    sentences=tokenized_sentences,
    vector_size=10,
    window=2,       # konteks 2 kata
    min_count=1,
    workers=1,
    seed=42
)
```

```python
# 4. Menampilkan seluruh vocabulary
print("\n----------------------------------------")
print("Vocabulary")
print("-" * 40)

for i, kata in enumerate(model.wv.index_to_key):
    print(f"{i+1}. {kata}")

# Menampilkan Semua Pasangan Target-Konteks
window = 2
pairs = []

for sentence in tokenized_sentences:
    for i, target in enumerate(sentence):
        # Menentukan batas konteks
        start = max(0, i - window)
        end = min(len(sentence), i + window + 1)
        # Mengambil pasangan target-konteks
        for j in range(start, end):
            # Jangan memasangkan kata dengan dirinya sendiri
            if i != j:
                context = sentence[j]
                pairs.append((target, context))
```

```python
# Menampilkan Hasil
print("\n----------------------------------------")
print("Pasangan (Target, Konteks):")
print("-" * 40)

for target, context in pairs:
    print(f"({target}, {context})")

print("\nJumlah pasangan:", len(pairs))
```

```python
# 5. Mengambil semua kata dan vektornya
kata = model.wv.index_to_key
vektor = model.wv.vectors

print("\n----------------------------------------")
print("Semua Vektor")
print("-" * 40)

for i, word in enumerate(kata):
    print(f"\nKata: {word}")
    print(f"Vektor: {vektor[i]}")
```

```python
# 6. Membuat DataFrame
df = pd.DataFrame(
    vektor,
    index=kata
)

df.columns = [
    f"dimensi_{i+1}"
    for i in range(model.vector_size)
]

print("\n----------------------------------------")
print("Tabel Representasi Vektor")
print("-" * 40)

print(df)
```

```python
# 7. Informasi vektor
print("\n----------------------------------------")
print("Informasi Vektor")
print("-" * 40)

print("Jumlah kata      :", len(kata))
print("Dimensi per kata :", model.vector_size)
print("Ukuran matriks   :", vektor.shape)
```
