### Pencarian dan Penambangan Web

Penambangan Web (PPW) merupakan mata kuliah di program studi Teknik Informatika Universitas Trunojoyo Madura (UTM) yang berfokus pada teknik ekstraksi, pemrosesan, dan analisis data berskala besar dari internet. Mahasiswa mempelajari cara mengumpulkan data secara otomatis menggunakan metode crawling dan scraping, sekaligus meletakkan dasar text processing untuk mengubah data mentah dari berbagai situs web menjadi struktur data yang siap diolah.

Pembahasan dilanjutkan dengan penerapan algoritma kecerdasan buatan, seperti Natural Language Processing (NLP), klasifikasi teks, clustering, dan analisis sentimen terhadap data web. Melalui praktikum dan pengerjaan proyek, mahasiswa Teknik Informatika UTM dilatih untuk membangun pipeline pemrosesan data otomatis yang mampu menarik wawasan (insight), tren, dan pengetahuan berharga untuk kebutuhan riset maupun industri.
